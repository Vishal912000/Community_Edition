```python
%Y  = Year (4 digits)       -> 2026
%y  = Year (2 digits)       -> 26
%m  = Month (01-12)         -> 05
%d  = Day (01-31)           -> 31
%H  = Hour (00-23)          -> 14
%M  = Minute (00-59)        -> 25
%S  = Second (00-59)        -> 30
%w  = Day of week (0-6)     -> 0 (Sunday)
%j  = Day of year (001-366) -> 151
%W  = Week of year          -> 21
%%  = Literal % character   -> %

Common combinations:
%Y-%m-%d          -> 2026-05-31
%d-%m-%Y          -> 31-05-2026
%Y-%m-%d %H:%M:%S -> 2026-05-31 14:25:30
%H:%M:%S          -> 14:25:30
```

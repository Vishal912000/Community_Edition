**Recursive CTE (Common Table Expression)** :-

A Recursive Common Table Expression (Recursive CTE) is a type of Common Table Expression (CTE) that references itself to repeatedly execute a query until a specified stopping condition is met. It is primarily used to retrieve and process hierarchical or recursive data, such as employee-manager relationships, organizational charts, folder structures, and parent-child records

In recursive cte we need to use **"recursive"** keyword.

In recursive cte we will have 
--> Anchor Query
--> Recursive query
--> Filter to stop recursion

--Write a recurive query to print number 1 to 10

```python
%sql

with recursive cte_numbers as (
    select 1 as num 

    union all 

    select num +1 from cte_numbers 
    where num <10 )

select * from cte_numbers 
```

```python
%sql
WITH recursive numbers AS (
    SELECT 1 AS num
    UNION ALL
    SELECT num + 1
    FROM numbers
    WHERE num < 10
)
SELECT *
FROM numbers;
```

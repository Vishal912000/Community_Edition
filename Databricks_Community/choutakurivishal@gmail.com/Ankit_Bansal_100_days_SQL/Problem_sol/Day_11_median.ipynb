What is Median?

Median is the middle value when data is arranged in ascending or descending order.

It represents the typical value in a dataset and is less affected by extreme values (outliers).

Why Not Use Average?

Suppose salaries are:

20,000
25,000
30,000
35,000
5,00,000
Average
(20,000 + 25,000 + 30,000 + 35,000 + 5,00,000) / 5

= 1,22,000

Average salary appears to be ₹1.22 lakh.

But most employees earn only ₹20k–₹35k.

Median
20,000  25,000  [30,000]  35,000  5,00,000

```python
%sql
select * from ankit_bansal_sql_100_days.days.emp_11
```

```python
%sql
with cte as (
select *,
row_number() over (order by emp_age asc) as rnk_asc,
row_number() over (order by emp_age desc) as rnk_desc
from   ankit_bansal_sql_100_days.days.emp_11
--where emp_id!=10
)

select avg(emp_age) from cte
where abs(rnk_asc-rnk_desc)<=1

```

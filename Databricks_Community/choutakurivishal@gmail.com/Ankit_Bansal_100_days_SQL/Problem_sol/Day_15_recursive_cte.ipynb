You are given a table that contains sales periods for products, along with average daily sales.

Each row represents:

A product

A start date

An end date

Average daily sales during that period

🧠 Problem Requirement

For each product, calculate total sales per year.

⚠️ Important catch:
A single period may span multiple years

Sales must be split year-wise

You cannot simply do:
DATEDIFF(period_end, period_start) * average_daily_sales
That only works when the entire period is within one year ❌


```python
%sql
select * from ankit_bansal_sql_100_days.days.sales
```

```python
%sql

with recursive r_cte MAX RECURSION LEVEL 1000 as (
    select min(period_start) as min_period ,max(period_end) as max_period
    from ankit_bansal_sql_100_days.days.sales

    union all 

    select DATE_ADD(min_period, 1) as min_period,max_period 
    from r_cte
    where min_period<max_period)
select product_id ,year(min_period) as report_year,sum(average_daily_sales) as total_amount from r_cte as r
join ankit_bansal_sql_100_days.days.sales s
on r.min_period between s.period_start and s.period_end
group by product_id,year(min_period)
order by product_id,year(min_period)
```

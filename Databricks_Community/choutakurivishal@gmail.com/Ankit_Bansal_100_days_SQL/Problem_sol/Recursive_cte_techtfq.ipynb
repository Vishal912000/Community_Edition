Find the hiearchy all employees reporting to a Bob(Top-Down)

```python
%sql
with recursive r_cte as 
(select emp_id,emp_name ,manager_id,designation
 from ankit_bansal_sql_100_days.days.employee_techtfq
 where emp_name='Bob'

 union all

 select e.emp_id,e.emp_name,e.manager_id,e.designation
 from r_cte  r 
 inner join ankit_bansal_sql_100_days.days.employee_techtfq  e
 on r.emp_id=e.manager_id
)

select * from r_cte

```

```python
%sql
with recursive cte as (
    select *,1 as level from ankit_bansal_sql_100_days.days.employee_techtfq
    where emp_name='Grace'

    union all 

    select e.*,level +1 from 
    cte c
    join ankit_bansal_sql_100_days.days.employee_techtfq e
    on c.manager_id=e.emp_id)

select *  from cte
```

```python
%sql


--reversing the level number if we want to see from CEO to emp using row number ,Same query as above only difference is rn
with recursive cte as (
    select *,1 as level from ankit_bansal_sql_100_days.days.employee_techtfq
    where emp_name='Grace'

    union all 

    select e.*,level +1 from 
    cte c
    join ankit_bansal_sql_100_days.days.employee_techtfq e
    on c.manager_id=e.emp_id)

select *,row_number() over ( order by level desc)  as rn  from cte
order by rn 
```

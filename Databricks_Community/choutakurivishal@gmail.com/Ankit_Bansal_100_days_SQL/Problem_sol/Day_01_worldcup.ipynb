```python
%sql
with cte as (select team_1 as team_name,case when team_1=winner then 1 else 0 end as win_flag from ankit_bansal_sql_100_days.days.icc_world_cup
union all
select team_2 as team_name ,case when team_2=winner then 1 else 0 end as win_flag from ankit_bansal_sql_100_days.days.icc_world_cup)

select team_name,count(*) as no_of_matches_played ,sum(win_flag) as no_of_matches_won ,count(*)-sum(win_flag) as no_of_losses 
from cte
group by team_name
order by no_of_matches_won desc
```

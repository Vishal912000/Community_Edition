```python
%sql
with resources as (
select concat_ws(', ', collect_list(distinct(resources))) as resources ,name   from ankit_bansal_sql_100_days.days.entries
group by name),

total_visits as (select name,count(*) as no_of_visits from ankit_bansal_sql_100_days.days.entries
group by name ),

most_visit as (select name,floor as most_visited_floor, rank() over(partition by name order by count(*) desc)  as rn from ankit_bansal_sql_100_days.days.entries
group by name,floor )

select mv.name,  mv.most_visited_floor,tv.no_of_visits,r.resources from most_visit mv 
join total_visits tv 
on mv.name=tv.name
join resources r
on mv.name=r.name
where mv.rn=1



```

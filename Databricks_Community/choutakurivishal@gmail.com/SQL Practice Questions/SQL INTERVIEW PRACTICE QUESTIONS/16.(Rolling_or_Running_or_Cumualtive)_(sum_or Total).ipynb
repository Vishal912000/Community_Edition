**A Rolling total (also called Running total or Cumulative sum)** is a continuously updated total where each row includes the current value plus all previous values based on a defined order (usually date, sequence, or transaction ID)

**Formula:-**

Running Total at row n:

Current Value + Sum of all prior values

```python
from pyspark.sql.types import StructField,StructType,IntegerType,StringType,DateType,TimestampType
from datetime import date,datetime
```

```python
schema=StructType(fields=[StructField("year",IntegerType(),nullable=True),
                        StructField("item",StringType(),nullable=True),
                        StructField("month",StringType(),nullable=True),
                        StructField("price",IntegerType(),nullable=True),
                        StructField("quantity",IntegerType(),nullable=True)
                            ])
```

```python
data=[(2020,"laptop","JAN",1000,7),
      (2021,"laptop","FEB",500,10),
      (2022,"laptop","MAR",300,5),
      (2023,"laptop","APR",10,6),
      (2024,"laptop","MAY",200,3),
      (2025,"laptop","JUN",50,4)
]
```

```python
df=spark.createDataFrame(data,schema)
display(df)
```

Rolling Sum/Cumulative Sum/Running Sum

```python
df.createOrReplaceTempView("df")
df_2=spark.sql(f"""select year,item,month,price,sum(price ) over(order by year)  as rolling_sum from df""")
display(df_2)
```

```python
#pyspark
from pyspark.sql.window import Window
from pyspark.sql import functions as F

window_spec=Window.orderBy(df.year )

df_py_1=df.withColumn("r_sum",F.sum(df.price).over(window_spec))
display(df_py_1)


```


Rolling window sum /moving  window sum 

```python
df_3 = spark.sql("""
    select year, item, month, price,
    sum(price) over (order by year rows between 2 preceding and current row) as rolling_window_sum
    from df
""")
display(df_3)
```

```python
#pyspark
from pyspark.sql.window import Window
from pyspark.sql import functions as F

window_spec_moving_sum = Window.orderBy(df.year).rowsBetween(-2, 0)

df_py_2 = df.withColumn("rolling_window_sum", F.sum(df.price).over(window_spec_moving_sum))
display(df_py_2)
```

```python
from pyspark.sql.types import StructType,StructField,IntegerType,StringType,DateType,TimestampType
from datetime import date,datetime
```

```python
schema=StructType(fields=[StructField("id",IntegerType(),nullable=True),
                          StructField("name",StringType(),nullable=True),
                          StructField("dept_id",IntegerType(),nullable=True),
                          StructField("salary",IntegerType(),nullable=True),
                          StructField("emp_age",IntegerType(),nullable=True),
                          StructField("d.o.b",DateType(),nullable=True),
                          StructField("login_time",TimestampType(),nullable=True)
                          ])
```

```python
data = [
    (1,  "Ankit",   100, 10000, 39, date(2025, 4, 1), datetime(2025, 4, 1, 9, 0, 0)),
    (2,  "Mohit",   100, 15000, 48, date(2025, 4, 1), datetime(2025, 4, 1, 9, 15, 0)),
    (3,  "Vikas",   100, 12000, 37, date(2025, 4, 1), datetime(2025, 4, 1, 9, 30, 0)),
    (4,  "Rohit",   100,  5000, 16, date(2025, 4, 1), datetime(2025, 4, 1, 9, 45, 0)),
    (5,  "Mudit",   200, 12000, 55, date(2025, 4, 1), datetime(2025, 4, 1, 10, 0, 0)),
    (6,  "Agam",    200, 10000, 14, date(2025, 4, 1), datetime(2025, 4, 1, 10, 15, 0)),
    (7,  "Sanjay",  200,  9000, 13, date(2025, 4, 1), datetime(2025, 4, 1, 10, 30, 0)),
    (8,  "Ashish",  300,  5000, 12, date(2025, 4, 1), datetime(2025, 4, 1, 10, 45, 0)),
    (9,  "Mukesh",  300,  6000, 51, date(2025, 4, 1), datetime(2025, 4, 1, 11, 0, 0)),
    (10, "Rakesh",  500,  7000, 50, date(2025, 4, 1), datetime(2025, 4, 1, 11, 15, 0))
]
```

```python
df=spark.createDataFrame(data,schema)
display(df)
```

```python
df.createOrReplaceTempView("df")
df_2=spark.sql(f"""with cte as (select id,name,dept_id,salary,avg(salary) over (partition by dept_id) as dept_avg_sal from df)
               select * from cte 
               where salary>dept_avg_sal""")
display(df_2)
```

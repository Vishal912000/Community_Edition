```python
%sql
select request_at,
count(status) ,
case when status like '%cancelled%' then 1 else 0 end as cancelled_flag
from ankit_bansal_sql_100_days.days.Trips t
left join  ankit_bansal_sql_100_days.days.Users u
on t.client_id=u.users_id
where u.banned!='Yes'
group by request_at,case when status like '%cancelled%' then 1 else 0 end
```

```python
%sql
select request_at,
count(status) as total_trips ,
sum(case when status like '%cancelled%' then 1 else 0 end) as cancelled_trip_count,
sum(case when status like '%cancelled%' then 1 else 0 end)/count(status) *100 as cancelled_percent
from ankit_bansal_sql_100_days.days.Trips t
left join  ankit_bansal_sql_100_days.days.Users u
on t.client_id=u.users_id
where u.banned!='Yes'
group by request_at
```

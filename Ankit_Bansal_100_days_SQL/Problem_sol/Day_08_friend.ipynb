```python
%sql
with  scores as(
select pid,sum(Score) as total_friend_score,count(*) as no_of_friends from ankit_bansal_sql_100_days.days.friend f
join  ankit_bansal_sql_100_days.days.person p on f.fid=p.person_id
group by pid
having sum(Score) >100)

select  s.pid,s.total_friend_score,s.no_of_friends,p.Name from scores as s
join ankit_bansal_sql_100_days.days.person  as p
on s.pid=p.person_id
```

```python
%sql
with scores as (
select first_player as player,first_score as score from ankit_bansal_sql_100_days.days.matches
union all 
select second_player as player,second_score as score from ankit_bansal_sql_100_days.days.matches),

cte as (select s.player,sum (s.score) as score ,p.group_id
from scores s
join  ankit_bansal_sql_100_days.days.players p
on s.player=p.player_id
group by player,p.group_id
order by p.group_id,s.player),

cte_2 as (select *, rank() over (partition by group_id order by score desc ,player asc ) as rnk from cte)

select * from cte_2 
where rnk=1
```
